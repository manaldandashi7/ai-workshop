## 🚀 Step 3 — Your First API Call

Let's send one message and see what comes back. This is the core of everything.

````python
from openai import OpenAI

# Create the client
client = OpenAI()

# Send a message
response = client.chat.completions.create(
    model="gpt-4o-mini",   # Fast and affordable — perfect for learning
    messages=[
        {
            "role": "user",
            "content": "Hello! Introduce yourself in one sentence."
        }
    ]
)

# Print the response
print(response.choices[0].message.content)
````

## 🧠 Step 5 — Multi-Turn Chatbot (with Memory)

Key insight: **LLMs have no memory**. Every call is independent.

To make a chatbot that "remembers", you send the **entire conversation history** each time.

```
Call 1: [system, user_msg_1]  →  bot_reply_1
Call 2: [system, user_msg_1, bot_reply_1, user_msg_2]  →  bot_reply_2
Call 3: [system, user_msg_1, bot_reply_1, user_msg_2, bot_reply_2, user_msg_3]  →  bot_reply_3
```

````python
# ─── Full multi-turn chatbot ───────────────────────────────────────────────

# 1. Start with a system prompt
messages = [
    {
        "role": "system",
        "content": (
            "You are a helpful and encouraging CS study assistant. "
            "Use analogies to explain concepts. Keep answers concise."
        )
    }
]

print("🤖 Study Bot ready! Type 'quit' to stop.\n")
print("-" * 50)

while True:
    # 2. Get user input
    user_input = input("You: ").strip()

    if user_input.lower() in ["quit", "exit", "bye", ""]:
        print("\n🤖 Goodbye! Keep coding! 💪")
        break

    # 3. Add user message to history
    messages.append({"role": "user", "content": user_input})

    # 4. Call API with full history
    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages,
        temperature=0.7,    # 0 = focused, 1 = creative
        max_tokens=300      # Limit response length
    )

    # 5. Extract reply
    reply = response.choices[0].message.content
    print(f"\n🤖 Bot: {reply}\n")
    print("-" * 50)

    # 6. Add bot reply to history (important!)
    messages.append({"role": "assistant", "content": reply})
````

## 🎁 Bonus — Useful Snippets

These are things you might need as you build more advanced bots.

````python
# ── Bonus 1: Limit conversation history to save tokens ────────────────────
# If you chat too long, the API call gets expensive. This keeps only recent messages.

MAX_HISTORY = 10

def trim_history(messages, max_history=10):
    """Keep system prompt + last max_history messages."""
    system = [messages[0]]  # Always keep the system prompt
    recent = messages[1:][-max_history:]  # Keep last N exchanges
    return system + recent

# Usage inside your loop:
# messages = trim_history(messages)

print("✅ trim_history() defined!")
````

## 🎨 UI Changes in This Version

In this version, we did not change the AI model or the chatbot logic.  
Instead, we improved only the **user interface (UI)** using Gradio.

````python
!pip install gradio -q

import gradio as gr
from openai import OpenAI

client = OpenAI()

# ─── SYSTEM PROMPT ─────────────────────────────────────────────
SYSTEM_PROMPT = "You are a friendly CS study assistant. Use analogies."

# ─── CHAT FUNCTION ─────────────────────────────────────────────
def chat(message, history):
    messages = [{"role": "system", "content": SYSTEM_PROMPT}]

    for h in history:
        messages.append({"role": "user", "content": h[0]})
        messages.append({"role": "assistant", "content": h[1]})

    messages.append({"role": "user", "content": message})

    response = client.chat.completions.create(
        model="gpt-4o-mini",
        messages=messages
    )

    return response.choices[0].message.content


# ─── MODERN DARK THEME ─────────────────────────────────────────
theme = gr.themes.Soft(
    primary_hue="indigo",
    secondary_hue="purple",
    neutral_hue="slate",
    radius_size="lg",
    font=[gr.themes.GoogleFont("Inter"), "system-ui", "sans-serif"]
)

# ─── CUSTOM CSS (chat-style feel) ──────────────────────────────
css = """
body {
    background: #0f172a;
}

.gradio-container {
    max-width: 900px !important;
    margin: auto;
}

h1, h2, h3 {
    color: white !important;
}

#chatbot {
    border-radius: 16px;
}

button {
    border-radius: 10px !important;
}
"""

# ─── LAUNCH UI ─────────────────────────────────────────────────
gr.ChatInterface(
    chat,
    title="🤖 CS Study Assistant",
    theme=theme,
    css=css
).launch(share=True)
````
