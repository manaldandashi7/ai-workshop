---
## ✍️ Step 2 — YOU write the training data

This is the most important step. **The model knows nothing right now.**

You're going to teach it by giving it examples — just like how you'd teach a child:
- *"This sentence is positive."*
- *"This sentence is negative."*

👇 **Add at least 5 examples of your own in each list below.**

````python
# ── YOUR TRAINING DATA ────────────────────────────────────────────────────
# Label 1 = Positive 😊   |   Label 0 = Negative 😠
# Add as many examples as you like — more data = smarter model!

positive_examples = [
    "I love this movie",
    "This is absolutely amazing",
    "What a fantastic experience",
    "I am so happy today",
    "This food is delicious",
    "Best day of my life",
    "I really enjoyed this",
    "Outstanding performance",
    "This made me smile",
    "Highly recommend it",
    "It as fun",
    "that is amazing",
    "happy time"
    # ← Add your own positive sentences here!
]

negative_examples = [
    "I hate this movie",
    "This is absolutely terrible",
    "What a horrible experience",
    "I am so disappointed",
    "This food is disgusting",
    "Worst day of my life",
    "I really disliked this",
    "Awful performance",
    "This made me angry",
    "Would not recommend it",
    "it sucks",
    "i am down"
    # ← Add your own negative sentences here!
]
````

🧩We combine all sentences into one list (texts) and create a matching list of labels (labels), so each sentence has a corresponding value 
(1 = positive, 0 = negative),we wnd up with two lists ready to be used in fit()


🔑An additional simple table of text + label is provivded to see how data are stored

````python
# ── Combine and label ─────────────────────────────────────────────────────
texts  = positive_examples + negative_examples
labels = [1] * len(positive_examples) + [0] * len(negative_examples)

print(f"📊 Training examples: {len(texts)} total")
print(f"   😊 Positive: {len(positive_examples)}")
print(f"   😠 Negative: {len(negative_examples)}")
````

````python
import pandas as pd

# Create a table (DataFrame)
df = pd.DataFrame({
    "Text": texts,
    "Label": labels
})

# Show first rows
df.head()
````

---
## 🔢 Step 3 — Turn words into numbers

Computers can't read words — they only understand numbers.

We use a technique called **TF-IDF** (Term Frequency–Inverse Document Frequency).
It converts each sentence into a list of numbers that represents which words appear in it.

Think of it like this:
> Each sentence becomes a "fingerprint" of numbers — words the model can actually work with.

````python
from sklearn.feature_extraction.text import TfidfVectorizer

# The vectorizer learns the vocabulary from our data
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(texts)  # Transform sentences → numbers

print(f"📐 Each sentence is now a vector of {X.shape[1]} numbers")
print(f"   (one number per unique word in your training data)")
print(f"\n📝 Words the model knows: {vectorizer.get_feature_names_out()[:20]} ...")
````

---
## 🏋️ Step 4 — TRAIN THE MODEL

This is the moment. We feed our labeled examples to a **Logistic Regression** model.

The model looks at all the number-fingerprints and finds a pattern:
> *"Sentences with 'love', 'amazing', 'fantastic' tend to be positive. Sentences with 'hate', 'terrible', 'awful' tend to be negative."*

Watch the training happen in real time 👇

# 👨‍💻Quick note
👉 When you call fit in scikit-learn, the model learns patterns that link the input data (texts) to the labels (1 = positive, 0 = negative).

````python
from sklearn.linear_model import LogisticRegression
import time

print("🏋️  Training started...")
start = time.time()

# Create and train the model
model = LogisticRegression(max_iter=1000)
model.fit(X, labels)

elapsed = time.time() - start
print(f"✅ Training complete in {elapsed:.3f} seconds!")
print(f"\n💡 The model has learned from {len(texts)} examples.")
print("   It's now ready to classify sentences it has NEVER seen before.")
````

---
## 🧪 Step 5 — Test it!

Now the fun part. Let's throw sentences at it that **it has never seen** during training.

Will it get them right? That depends on how good YOUR training data was 😄

````python
def predict(sentence):
    """Run a sentence through the trained model and show the result."""
    vec = vectorizer.transform([sentence])
    prediction = model.predict(vec)[0]
    confidence = model.predict_proba(vec)[0]

    label = "😊 POSITIVE" if prediction == 1 else "😠 NEGATIVE"
    conf  = confidence[prediction] * 100

    print(f"  Input      : \"{sentence}\"")
    print(f"  Prediction : {label}")
    print(f"  Confidence : {conf:.1f}%")
    print()

print("=" * 55)
print("           🧪 TESTING THE MODEL")
print("=" * 55)
print()

# Try these — feel free to change them!
test_sentences = [
    "This was a wonderful experience",    # Should be positive
    "I absolutely hated every second",    # Should be negative
    "Not bad at all",                     # Tricky!
    "Could have been better",             # Tricky!
    "I love how terrible this is",        # Very tricky — sarcasm!
]

for s in test_sentences:
    predict(s)
````

## 📊 Bonus — See what the model actually learned

Which words push the model toward positive? Which toward negative?
This is what's happening *inside* the model.

````python
import numpy as np

feature_names = vectorizer.get_feature_names_out()
coefficients  = model.coef_[0]

# Top words for each class
top_n = 8
top_positive_idx = np.argsort(coefficients)[-top_n:][::-1]
top_negative_idx = np.argsort(coefficients)[:top_n]

print("🧠 What the model learned:\n")
print("😊 Words that STRONGLY suggest POSITIVE:")
for i in top_positive_idx:
    bar = "█" * int(coefficients[i] * 5)
    print(f"   {feature_names[i]:<20} {bar}  ({coefficients[i]:.2f})")

print()
print("😠 Words that STRONGLY suggest NEGATIVE:")
for i in top_negative_idx:
    bar = "█" * int(abs(coefficients[i]) * 5)
    print(f"   {feature_names[i]:<20} {bar}  ({coefficients[i]:.2f})")
````
